#### When to use dropna()

Usually when:

- the dataset is huge,
- only a few rows have missing values,
- and removing those rows won't hurt the analysis.

Otherwise, we look for smarter approaches, like:

- filling missing values,
- removing only specific columns,
- or removing rows only when certain columns are missing.

### Choosing the right filling method

| Method        | Best for                          | Example                         |
| ------------- | --------------------------------- | ------------------------------- |
| Constant      | Placeholder values                | Replace with 0 or "Unknown"     |
| Mean          | Normally distributed numeric data | Height, temperature             |
| Median        | Numeric data with outliers        | Income, house prices            |
| Mode          | Categorical data                  | Country, gender, weather type   |
| Forward fill  | Time series                       | Sensor readings, stock prices   |
| Backward fill | Time series                       | Missing values at the beginning |
| Interpolation | Continuous time series            | Temperature, rainfall           |
