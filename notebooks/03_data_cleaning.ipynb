| Code                    | Purpose                                              |
| ----------------------- | ---------------------------------------------------- |
| `df.isna()`             | Show every missing value as True/False               |
| `df.isna().sum()`       | Count missing values in each column                  |
| `df.isna().sum(axis=1)` | Count missing values in each row                     |
| `df.isna().any()`       | Check which columns contain missing values           |
| `df.isna().any().any()` | Check if the entire DataFrame has any missing values |

#### When to use dropna()

Usually when:

- the dataset is huge,
- only a few rows have missing values,
- and removing those rows won't hurt the analysis.

Otherwise, we look for smarter approaches, like:

- filling missing values,
- removing only specific columns,
- or removing rows only when certain columns are missing.

```python
temp_avg_year_country.dropna(how="any").shape
```

```python
temp_avg_year_country.dropna(how="all").shape
```

```python
temp_avg_year_country
```

```python
# Removing duplicates in a specific column using subset.

temp_avg_year_country.dropna(subset=[1984])
```

#### DropNA Challenge 1

A climate researcher says:

> "I'm studying long-term climate trends, so I only want countries that have a complete temperature record for every year in the dataset."

Using `temp_avg_year_country`, produce a DataFrame that satisfies this request.

```python
temp_avg_year_country.dropna(how="any")
```

#### Data Analysis Question

A researcher now changes the requirement.

> "I don't mind if a country is missing a few years of data. I only want to remove countries that have **no temperature records whatsoever**."

Using `temp_avg_year_country`, produce the required DataFrame.

```python
temp_avg_year_country.dropna(axis= 0, how="all")
```

#### Data Analysis Question

A climate researcher says:

> "For my study, I only care about the years **2000**, **2001**, and **2002**.

> Remove countries that are missing temperature data in **any** of those three years, but ignore missing values in every other year."

Produce the required DataFrame.

```python
temp_avg_year_country.dropna(subset=[2000,2001,2002])
```

#### fillna()

A climate researcher says:

"I don't want to remove any countries from my pivot table.

Instead, replace every missing temperature value with 0 while keeping all countries and all years."

Use temp_avg_year_country to produce the required DataFrame.

```python
temp_avg_year_country.fillna(0)
```

A researcher says:

"I don't want to replace missing temperatures with zero because that would be misleading.

Instead, replace every missing value with the average temperature of the entire pivot table column (each year's average temperature)."

```python
temp_avg_year_country.fillna(temp_avg_year_country.mean())
```


A researcher believes the temperature data contains a few unusually hot years that could affect the average.

Instead of filling missing values with the **mean**, replace every missing value in `temp_avg_year_country` with the **median temperature of its respective year**.

```python
temp_avg_year_country.fillna(temp_avg_year_country.median())
```
