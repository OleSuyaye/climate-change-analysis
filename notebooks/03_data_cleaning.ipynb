#### When to use dropna()

Usually when:

- the dataset is huge,
- only a few rows have missing values,
- and removing those rows won't hurt the analysis.

Otherwise, we look for smarter approaches, like:

- filling missing values,
- removing only specific columns,
- or removing rows only when certain columns are missing.

```python
temp_avg_year_country.dropna(how="any").shape
```

```python
temp_avg_year_country.dropna(how="all").shape
```

```python
temp_avg_year_country
```

```python
# Removing nulls in a specific column using subset (Removes the whole row)

temp_avg_year_country.dropna(subset=[1984])
```

#### fillna()

A climate researcher says:

"I don't want to remove any countries from my pivot table.

Instead, replace every missing temperature value with 0 while keeping all countries and all years."

Use temp_avg_year_country to produce the required DataFrame.

```python
temp_avg_year_country.fillna(0)
```

A researcher says:

"I don't want to replace missing temperatures with zero because that would be misleading.

Instead, replace every missing value with the average temperature of the entire pivot table column (each year's average temperature)."

```python
temp_avg_year_country.fillna(temp_avg_year_country.mean())
```


A researcher believes the temperature data contains a few unusually hot years that could affect the average.

Instead of filling missing values with the **mean**, replace every missing value in `temp_avg_year_country` with the **median temperature of its respective year**.

```python
temp_avg_year_country.fillna(temp_avg_year_country.median())
```

Challenge 1: The Cold Drop 

You're inspecting temp_avg_year_country. You notice that some countries have missing average temperatures for certain years. Your lead data scientist asks you to completely remove any country (row) that has at least one missing value across any of the years, and save this clean view as a new variable called complete_countries_df.

Write the code to accomplish this.

```python
complete_countries_df= temp_avg_year_country.dropna()
complete_countries_df
```

Challenge 2: The Timeline Gap 

Now, instead of dropping entire countries, your team wants to keep all countries, but deal with missing values year-by-year.

A senior analyst asks: "If a country is missing its average temperature for a specific year, drop any year (column) from the pivot table that contains any missing data across all countries." Save this as complete_years_df.

```python
complete_years_df = temp_avg_year_country.dropna(axis = 1)
complete_years_df
```

Challenge 3: The Reasonable Estimate 

Now, let's swap from dropping data to filling missing data!

Your lead analyst comes to you with a compromise:

"We don't want to lose whole rows or columns. Instead, if a country is missing its average temperature for any given year, let's fill in that missing cell using the overall average temperature of the entire dataset."

Save this filled table as filled_temp_df.

```python
overall_mean = temp_avg_year_country.mean().mean()
overall_mean

filled_temp_df = temp_avg_year_country.fillna(overall_mean)
filled_temp_df
```

Challenge 4: Forward Fill Across Time

Save this as ffill_temp_df.

(Hint: Try using .ffill() or .fillna(method='ffill'), but remember to specify the right axis so it moves across columns/years instead of down rows!)

Write the code to accomplish this.

```python
temp_avg_year_country
```

```python
ffill_temp_df = temp_avg_year_country.ffill(axis=1)
ffill_temp_df
```

 Challenges 5

 Modify your code so that it first tries to forward fill across years (axis=1), and then immediately uses backward fill (.bfill()) across years (axis=1) to catch any remaining leading missing values. Save the result as fully_filled_df.

### Boolean Filtering 
Revision and reminder

```python
climate_df[
    climate_df["global_avg_temperature"] > 25
]
```

```python
climate_df[
    climate_df["country"] == "Brazil"
    ].head(10).sort_values(["country"])
```

Suppose your manager asks:

Show me all records for Brazil after 2010.

```python
climate_df[(climate_df["country"] == "Brazil") & (climate_df["year"]>= 2010)].sort_values(["year"])
```

Suppose the manager says:

Show me records from Brazil OR Canada.

```python
brazil_canada_data_df = climate_df[(climate_df["country"]== "Brazil") | (climate_df["country"] == "Canada")]
brazil_canada_data_df 
```

Suppose the manager says:

Show me every country except Brazil.

```python
climate_df[climate_df["country"] != "Brazil"]
```

```python
climate_df[~(climate_df["country"] == "Brazil")]
```

isin()


```python
climate_df[climate_df["country"].isin(["Brazil", "Canada"])]
```

Combining isin() with AND

Suppose the manager says:

Show me Brazil and Canada after 2015.

```python
climate_df[~(climate_df["country"].isin(["Brazil", "Canada"])) & (climate_df["year"] >= 2015)]
```

between()

Suppose your manager asks:

Show me all climate records between 2000 and 2010.

```python
climate_df[climate_df["year"].between(2000,2010)].sort_values(["year"])
```
