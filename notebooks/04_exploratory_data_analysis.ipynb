# What is Exploratory Data Analysis (EDA)?

Before building dashboards or machine learning models, analysts ask:

"What story is the data telling me?"

EDA is the process of exploring a dataset to:

- Understand its structure
- Discover patterns
- Find trends
- Detect anomalies (outliers)
- Generate hypotheses
- Answer business questions

Think of EDA as investigation, not cleaning.

Cleaning fixes the data.

EDA understands the data.

Imagine a climate scientist asks:

"What is the average global temperature for each country?"
