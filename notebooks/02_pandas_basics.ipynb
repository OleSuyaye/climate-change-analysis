# Lesson 5: Filtering Rows With Multiple Conditions

# Lesson 8: Group by and Pivot Tables


```python
# Single column grouping (returns a series)
climate_df.groupby("country")["annual_rainfall_mm"].mean()
```

```python
# Multiple columns grouping (returns a dataframe)
climate_df.groupby("country")[["global_avg_temperature", "annual_rainfall_mm"]].mean()
```

### Challenges

### GroupBy Challenge 1

Using `groupby()`:

- Calculate the average (`mean`) **global average temperature** for each country.
- Store the result in a variable called `country_avg_temp`.
- Sort the results from **highest average temperature to lowest**.
- Display the top **5** countries.

Requirements:
- Do not use `pivot_table()`.
- Do not use `.agg()` for this challenge.
- Use only `groupby()`, `mean()`, `sort_values()`, and `head()`.

```python
country_avg_temp = climate_df.groupby("country")["global_avg_temperature"].mean()
country_avg_temp.sort_values(ascending=False).head(5)
```

### GroupBy Challenge 2

Using `groupby()`:

- Calculate the **maximum annual rainfall** for each country.
- Store the result in a variable called `country_max_rainfall`.
- Display the countries from **highest rainfall to lowest rainfall**.
- Show **all** countries.

Requirements:
- Do not use `.agg()`.
- Do not use `pivot_table()`.
- Use only `groupby()`, `max()`, and `sort_values()`.

```python
country_max_rainfall = climate_df.groupby("country")["annual_rainfall_mm"].max()
country_max_rainfall.sort_values(ascending=False)
```

### GroupBy Challenge 3 (Final)

Without using `.agg()`:

Create a summary that shows, for each country:

- The **minimum** global average temperature.
- Sort the results from the **coldest country** to the **warmest country** (based on that minimum temperature).
- Display only the **top 3** countries.

Requirements:
- Use only `groupby()`, `min()`, `sort_values()`, and `head()`.

```python
cold_to_warm_countries = climate_df.groupby("country")["global_avg_temperature"].min()
cold_to_warm_countries.sort_values().head(3)
```

## Pivot Tables

Create a pivot table that shows:

- Rows = Country
- Columns = Year
- Values = Global Average Temperature
- Aggregation = Mean

```python
climate_df.pivot_table(
    index = "country",
    values= "global_avg_temperature",
    columns= "year",
    aggfunc= "mean"
).head()
```

### Pivot Table Challenge 1

Create a pivot table that shows:

- Rows as `country`
- Columns as `year`
- Values as `annual_rainfall_mm`
- Display the **average annual rainfall** for each country in each year.

Requirements:
- Use `pivot_table()`.
- Do not use `groupby()`.
- Store the result in a variable called `country_year_rainfall`.

```python
country_year_rainfall = climate_df.pivot_table(
    index = "country",
    columns= "year",
    values = "annual_rainfall_mm",
    aggfunc= "mean"
)

country_year_rainfall
```

### Data Analysis Question

A climate researcher wants to compare how **global average temperatures** have changed over the years for every country.

Create a table that allows the researcher to quickly compare each country's average global temperature across all available years.

```python
yearly_temp_change = climate_df.pivot_table(
    index = "country",
    columns = "year",
    values = "global_avg_temperature",
    aggfunc= "mean"
)

yearly_temp_change
```

### Data Analysis Question

A climate policy team wants to identify years that experienced the highest temperatures in each country.

Create a table that makes it easy to see the **highest recorded maximum temperature** for every country in every year.

```python
yearly_max_temp = climate_df.pivot_table(
    index = "country",
    columns = "year",
    values = "max_temperature",
    aggfunc= "max"
)

yearly_max_temp
```
