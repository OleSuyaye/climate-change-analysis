# Lesson 5: Filtering Rows With Multiple Conditions

```python
climate_df.columns
```

```python
climate_df["global_avg_temperature"] > 20
```

```python
climate_df["co2_concentration_ppm"] > 400
```

```python
# Combining conditions using &
climate_df[(climate_df["global_avg_temperature"] > 16) & (climate_df["co2_concentration_ppm"] > 400)]
```

```python
# Using | for OR
climate_df[(climate_df["country"] == "USA") | (climate_df["country"] == "Germany")]
```

```python
# Using ~ for NOT
climate_df[~(climate_df["country"] == "USA")]

```

# Lesson 7: Creating New Columns

```python
climate_df.head(20)
```

```python
# Creating a new column

climate_df["doubled_temp"] = climate_df["global_avg_temperature"] * 2
climate_df
```

```python
# Rounding of global_avg_temperature to one decimal place into a new column
climate_df["rounded_global_avg_temp"] = climate_df["global_avg_temperature"].round(1)
climate_df
```

```python
# Creating a  new column for temperatures under 2 as cold

climate_df["is_cold"] = climate_df["max_temperature"] < 22
climate_df[climate_df["is_cold"] == True]
```
